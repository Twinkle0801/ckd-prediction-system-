<h2>Standardize categorical text</h2>

```python
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.replace('\t', '', regex=False)
    df[col] = df[col].replace('?', pd.NA)
```

```python
df[cat_cols].apply(lambda c: c.unique())
```

<h2>Fix dtypes</h2>

```python
numeric_cols = ['age','bp','sg','al','su','bgr','bu','sc','sod','pot',
                 'hemo','pcv','wc','rc']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
print(df.dtypes)
```

```python
df[numeric_cols].isnull().sum()
```

<h2> Impute missing values contextually</h2>

```python
from sklearn.impute import KNNImputer
```

```python
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```
